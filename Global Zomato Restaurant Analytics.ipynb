# Top 3 countries that uses zomato

```python
country_name = df_final['Country'].value_counts().index
```

```python
country_name
```

```python
df_final['Country'].value_counts()
```

```python
country_value = df_final['Country'].value_counts().values
country_value
```

```python
#visualizing top3
```

```python
plt.pie(country_value[:3],labels=country_name[:3],autopct='%0.01f%%',colors=['green','red','orange'],radius = 1,explode=[0,0.2,0])
plt.legend(loc=0)
plt.title('Top 3 countries that uses zomato',loc = None)
```

### Analysis of zomoto ratings

```python
df_final['Aggregate rating']
```

```python
df_final['Rating color']
```

```python
df_final['Rating text']
```

```python
rating = df_final.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:"Rating count"})
rating
```

```python
plt.figure(figsize=[12,6])
sns.barplot(x = 'Aggregate rating',y = 'Rating count',data=rating,palette="pastel")
```

```python
# create bar graph on basis of rating colors:
sns.barplot(x = 'Aggregate rating',y = 'Rating count',data=rating,hue="Rating color",palette=['black','red','orange','yellow','green','pink'])
```
